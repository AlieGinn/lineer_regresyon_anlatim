```python
import numpy as np
x = [20,30,40]
y = [3000,5000,10000]

ones_column = np.ones((len(x), 1))
print(ones_column)
```

```python

x_t=np.array(x).reshape(-1, 1)
print(x_t)
```

```python

X_Transformed=np.hstack((ones_column,x_t))
print(X_Transformed)
```

```python

x_relation = X_Transformed.T @ X_Transformed
x_relation
```

```python

x_y_relation = X_Transformed.T @ y
x_y_relation
```

```python
coefficiants=np.linalg.inv(x_relation) @ x_y_relation
coefficiants
```

```python
newY=[]
for i in x :
    sonuc=coefficiants[0]+coefficiants[1]*i
    newY.append(sonuc)
newY
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Veriler
x = [20, 30, 40,80]
y = np.array([3000, 5000, 10000,15000]) 

#kapalı çözüm
ones_column = np.ones((len(x), 1))
x_t = np.array(x).reshape(-1, 1)
X_Transformed = np.hstack((ones_column, x_t))


X_T_X = X_Transformed.T @ X_Transformed
X_T_y = X_Transformed.T @ y


coefficients = np.linalg.inv(X_T_X) @ X_T_y


newY = coefficients[0] + coefficients[1] * np.array(x)


print("katsayılar:", coefficients)
print("tahmin Y değerleri :", newY)
```

```python
# Görselleştirme
plt.figure(figsize=(10, 6)) 
plt.scatter(x, y, color='blue', label='Gerçek Veriler') 
plt.plot(x, newY, color='red', label='Tahmin Edilen Doğru')  
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Veri
X = np.array([20, 30, 40])
y = np.array([3000, 5000, 10000])

# Gradyan İniş
# Hiperparametreler
alpha = 0.00001  # Öğrenme oranı
iterations = 200   # Iterasyon sayısı


beta0, beta1 = 0, 0


for i in range(iterations):
    # Tahminler
    y_pred = beta0 + beta1 * X
    
   
    grad_beta0 = -2 * np.sum(y - y_pred)
    grad_beta1 = -2 * np.sum(X * (y - y_pred))
    
  
    beta0 = beta0 - alpha * grad_beta0
    beta1 = beta1 - alpha * grad_beta1
    
    # Her 10 iterasyonda bir sonucu yazdır
    if i % 10 == 0:
        print(f"Iterasyon {i}: β0 = {beta0:.2f}, β1 = {beta1:.2f}")

print(f"Final Katsayılar: β0 = {beta0:.2f}, β1 = {beta1:.2f}")
```
